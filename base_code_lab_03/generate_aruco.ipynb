```python
# Generate aruco markers and save them as images.

import cv2
import numpy as np
from cv2 import aruco


# Number of markers to generate
num_markers = 1
# Size of the marker images (in inches)
marker_size_inch = 4

# Create a dictionary of aruco markers
aruco_dict = aruco.getPredefinedDictionary(aruco.DICT_4X4_50)
marker_images = []

# Generate and save the markers
for i in range(num_markers):
    marker_image = aruco.generateImageMarker(aruco_dict, 7 + i, 1000)
    marker_image = cv2.cvtColor(marker_image, cv2.COLOR_GRAY2BGR)  # Convert to BGR format for saving
    # Save image
    cv2.imwrite(f'marker_{i}.png', marker_image)
    
    marker_images.append(marker_image)
```

```python
# Add images to a pdf of Letter size and make the images marker_size_inch x marker_size_inch
# One image per page
from fpdf import FPDF
import tempfile
import os

pdf = FPDF(orientation='P', unit='mm', format='Letter')
letter_width = 216  # 8.5 inches * 25.4 mm/inch ≈ 215.9, rounded to 216
letter_height = 279  # 11 inches * 25.4 mm/inch ≈ 279.4, rounded to 279
for marker_image in marker_images:
    pdf.add_page()
    marker_size_mm = marker_size_inch * 25.4
    x_offset = (letter_width - marker_size_mm) / 2
    y_offset = (letter_height - marker_size_mm) / 2
    with tempfile.NamedTemporaryFile(suffix='.png', delete=False) as tmp_file:
        cv2.imwrite(tmp_file.name, marker_image)
        pdf.image(tmp_file.name, x=x_offset, y=y_offset, w=marker_size_mm, h=marker_size_mm)
        os.unlink(tmp_file.name)
    
pdf.output("aruco_markers.pdf")
```
