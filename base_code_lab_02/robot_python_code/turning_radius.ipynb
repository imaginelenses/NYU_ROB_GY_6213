```python
a = 37.5
b = 38
c = 13.5
```

```python
import math

def triangle_circumcircle_radius(a, b, c):
    # Check if lengths form a triangle using triangle inequality
    if (a + b <= c) or (a + c <= b) or (b + c <= a):
        return None, "These lengths do not form a triangle."
    
    # Semi-perimeter
    s = (a + b + c) / 2
    
    # Area using Heron's formula
    area = math.sqrt(s * (s - a) * (s - b) * (s - c))
    
    # Circumradius formula: R = (a*b*c) / (4*area)
    R = (a * b * c) / (4 * area)
    
    return R, "These lengths form a triangle."

radius, message = triangle_circumcircle_radius(a, b, c)

print(message)
if radius:
    print(f"Circumcircle radius: {radius:.2f}")
```
